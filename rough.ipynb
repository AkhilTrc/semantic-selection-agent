```python
import json
import csv
import glob
import os

# Path to the folder containing your .jsonl files
input_folder = 'without'
output_folder = 'dynamic_temp_inventory_size'
os.makedirs(output_folder, exist_ok=True)

# Find all .jsonl files in the input folder
jsonl_files = sorted(glob.glob(os.path.join(input_folder, '*.jsonl')))

for filename in jsonl_files:
    inventory_sizes = []
    # Open and read each line in the .jsonl file
    with open(filename, 'r', encoding='utf-8') as infile:
        for line in infile:
            entry = json.loads(line)
            inventory_sizes.append(entry['inventory_size'])

    # Write to corresponding .csv in the output folder
    base = os.path.splitext(os.path.basename(filename))[0]
    csv_path = os.path.join(output_folder, f"{base}.csv")
    with open(csv_path, 'w', newline='', encoding='utf-8') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['inventory_sizes'])
        for value in inventory_sizes:
            writer.writerow([value])

    print(f"Saved: {csv_path}")
```
